```python
import pandas as pd

# 1. Recreate the dataset exactly from your image table
data = {
    'Toss': ['L', 'W', 'W', 'L', 'W', 'W', 'L'],
    'Temp': ['HOT', 'MILD', 'HOT', 'HOT', 'COOL', 'MILD', 'COOL'],
    'Venue': ['MUMBAI', 'DELHI', 'CHENNAI', 'MUMBAI', 'JAIPUR', 'CHENNAI', 'MUMBAI'],
    'Match': ['W', 'W', 'L', 'L', 'W', 'W', 'L']
}
df = pd.DataFrame(data)

# 2. Define your prediction request parameters
query = {
    'Toss': 'L',
    'Temp': 'COOL',
    'Venue': 'JAIPUR'
}

total_records = len(df)
classes = df['Match'].unique()  # ['W', 'L']
raw_scores = {}

# 3. Calculate Smoothed Naive Bayes Probabilities
for c in classes:
    df_class = df[df['Match'] == c]

    # Calculate Prior: P(Class)
    prior = len(df_class) / total_records

    # Calculate Likelihood: P(F1|C) * P(F2|C) * ...
    likelihood = 1.0
    for feature, value in query.items():
        # Count target occurrences
        feature_count = len(df_class[df_class[feature] == value])

        # Apply Laplace Smoothing to handle unseen/zero combinations gracefully
        unique_vals_count = len(df[feature].unique())
        conditional_prob = (feature_count + 1) / (len(df_class) + unique_vals_count)
        likelihood *= conditional_prob

    raw_scores[c] = prior * likelihood

# 4. Normalize scores into a total 100% distribution range
total_score = sum(raw_scores.values())
probabilities = {c: score / total_score for c, score in raw_scores.items()}

# 5. Output Prediction Summary
print("--- Naive Bayes Match Prediction ---")
for c, prob in probabilities.items():
    print(f"Probability of Match Result '{c}': {prob * 100:.2f}%")

final_prediction = max(probabilities, key=probabilities.get)
print(f"\nFinal Predicted Outcome: {final_prediction}")
```
