```python
 #1.Bascic DataFrame Operations
df=pd.read_csv("/content/drive/MyDrive/24UAM077_ML/EX01/tested.csv")
print("-------------Dataset:------------")
print(df)
print("-------------Dataset Head:------------")
print(df.head())
print("-------------Dataset Tail:------------")
print(df.tail())

print("-------------Dataset Describe:------------")
print(df.describe())
print("-------------Dataset Info:------------")
print(df.info())
print("-------------Columns:------------")
print(df.columns)


# 2.Filtering Dataset (Droping Features)
print("------------Data Set after Dropping Feature : PassengerId and cabin--------------")
df = df.drop(columns=['PassengerId','Name','Ticket','Cabin'])
print(df)

# 4.Handaling Missing Values
print("-------------Missing Values:------------")
print(df.isnull().sum())

embarked_imputer = SimpleImputer(strategy='most_frequent')
df[['Embarked']] = embarked_imputer.fit_transform(df[['Embarked']])

age_imputer = SimpleImputer(strategy='median')
df[['Age']] = age_imputer.fit_transform(df[['Age']])

print (df)

# 3.Handaling non-numberic values ,One hot encoading (on x) Lable encoading (on y)
X = df.drop(columns=['Survived'])
y = df['Survived']
X = pd.get_dummies(X, columns=['Sex', 'Embarked'])
print("-------------Data Set after One Hot Encoding:------------")
print(X)


# 5.Normalization (Min-Max Scaling)
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(X)
print(scaled_data)

#6. splitting the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    scaled_data,
    y,
    test_size=0.2,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```
